```polyglot-notebook
int[] numbers = { 5, 8, 1, 3, 9, 6, 7, 2 };

int min = numbers.Min();
int max = numbers.Max();
Console.WriteLine($"Минимальное значение: {min}");
Console.WriteLine($"Максимальное значение: {max}");

int sum = numbers.Sum();
Console.WriteLine($"Сумма всех элементов: {sum}");

double average = numbers.Average();
Console.WriteLine($"Среднее арифметическое: {average}");

var sortedNumbers = numbers.OrderBy(n => n).ToArray();
Console.WriteLine("Отсортированный массив: " + string.Join(", ", sortedNumbers));

int searchNumber = 9;
bool contains = numbers.Contains(searchNumber);
Console.WriteLine($"Содержит ли массив число {searchNumber}? {contains}");
```
