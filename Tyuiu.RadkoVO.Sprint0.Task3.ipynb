```polyglot-notebook
using System;

public class Subscription
{
    public string SubscriptionId { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }

    public Subscription(string subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription(int months)
    {
        Console.WriteLine($"Подписка '{ServiceName}' продлена на {months} мес. без дополнительных условий.");
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID подписки: {SubscriptionId}");
        Console.WriteLine($"Услуга: {ServiceName}");
        Console.WriteLine($"Стоимость: {Cost} руб.");
        Console.WriteLine($"Ежемесячная стоимость: {CalculateMonthlyCost()} руб.");
    }
}
```

```polyglot-notebook
public class OnlineServiceSubscription : Subscription
{
    public int MaxUsers { get; set; }

    public OnlineServiceSubscription(string id, string name, decimal cost, int maxUsers)
        : base(id, name, cost)
    {
        MaxUsers = maxUsers;
    }

    public override decimal CalculateMonthlyCost()
    {
        return Cost + (MaxUsers - 1) * 100;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine("Количество пользователей: " + MaxUsers);
    }
}
```

```polyglot-notebook
public class StreamingSubscription : Subscription
{
    public int MaxStreams { get; set; }

    public StreamingSubscription(string id, string name, decimal cost, int maxStreams)
        : base(id, name, cost)
    {
        MaxStreams = maxStreams;
    }

    public override void ExtendSubscription(int months)
    {
        Console.WriteLine("Продление стриминговой подписки '" + ServiceName + "' на " + months + " мес.");
        Console.WriteLine("Спецпредложение: +1 месяц бесплатно при продлении от 3 месяцев!");
        if (months >= 3)
        {
            Console.WriteLine("Вам начислен 1 бонусный месяц!");
        }
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine("Одновременных потоков: " + MaxStreams);
    }
}
```

```polyglot-notebook
public class VideoSubscription : Subscription
{
    public string VideoQuality { get; set; }

    public VideoSubscription(string id, string name, decimal cost, string videoQuality)
        : base(id, name, cost)
    {
        VideoQuality = videoQuality;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine("Качество видео: " + VideoQuality);
    }
}
```

```polyglot-notebook
OnlineServiceSubscription online = new OnlineServiceSubscription("SUB-001", "Онлайн-сервис", 500, 3);
StreamingSubscription streaming = new StreamingSubscription("SUB-002", "Стриминг", 800, 2);
VideoSubscription video = new VideoSubscription("SUB-003", "Видео", 600, "4K");

Subscription[] subscriptions = { online, streaming, video };

Console.WriteLine("=== Детали подписок ===");
foreach (var sub in subscriptions)
{
    sub.GetSubscriptionDetails();
    Console.WriteLine("---");
}

Console.WriteLine("=== Расчет ежемесячной стоимости ===");
foreach (var sub in subscriptions)
{
    Console.WriteLine(sub.ServiceName + ": " + sub.CalculateMonthlyCost() + " руб.");
}

Console.WriteLine("\n=== Продление подписок ===");
foreach (var sub in subscriptions)
{
    sub.ExtendSubscription(4);
    Console.WriteLine("---");
}
```
