```polyglot-notebook
class Animal
{
    private string name;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public void Eat()
    {
        Console.WriteLine("Животное ест...");
    }
}

class Cat : Animal
{
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myCat.Name = "Мурка";
Console.WriteLine(myCat.Name);
myCat.Eat();

myAnimal.Name = "Барсик";
Console.WriteLine(myAnimal.Name);
myAnimal.Eat();
```

```polyglot-notebook
class Animal
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }
}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Кошка мяукает");
    }
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myAnimal.MakeSound();
myCat.MakeSound();
```

```polyglot-notebook
abstract class Animal
{
    public abstract void MakeSound();

    public void Sleep()
    {
        Console.WriteLine("Животное спит...");
    }
}

class Dog : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Собака лает: Гав-гав!");
    }
}

Dog myDog = new Dog();

myDog.MakeSound();
myDog.Sleep();
```

```polyglot-notebook
using System;

public class Heart
{
    public void Beat() => Console.WriteLine("Сердце бьётся");
    public void Stop() => Console.WriteLine("Сердце остановилось");
}

public class Animal
{
    private Heart heart;

    public Animal(Heart heart)
    {
        this.heart = heart;
    }

    public void Live()
    {
        heart.Beat();
        Console.WriteLine("Животное живёт");
        heart.Stop();
    }
}

Heart myHeart = new Heart();
Animal myAnimal = new Animal(myHeart);

myAnimal.Live();
```

```polyglot-notebook
using System;

abstract class Animal
{
    public abstract void Move();
}

class Dog : Animal
{
    private Owner owner;

    public Dog(Owner owner)
    {
        this.owner = owner;
    }

    public override void Move()
    {
        Console.WriteLine("Собака бежит...");
    }
}

class Owner
{
    public string Name { get; set; }

    public Owner(string name)
    {
        Name = name;
    }

    public void Walk(Animal animal)
    {
        Console.WriteLine($"{Name} гуляет с питомцем...");
        animal.Move();
    }
}

Owner Igor = new Owner("Игорь");
Dog myDog = new Dog(Igor);

myDog.Move();
Igor.Walk(myDog);
```
